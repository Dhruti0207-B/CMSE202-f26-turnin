&#9989; **Question 4.1 (4 points)**: Complete the `Recipe` class in the cell below.

1. `__init__` takes the recipe's `name` and the number of `servings` it makes. Store both as attributes, and create an attribute `ingredients` that starts as an empty dictionary.
2. `add_ingredient` takes an `ingredient` (a string such as `"spaghetti"`, which you use as the key in `ingredients`) and an `amount`. If the ingredient is already in the recipe, add `amount` to its current amount. Otherwise, add the ingredient to the dictionary with that amount.

Then run the test cell below it.

````python
### Put your code here
class Recipe:
    def __init__(self, name, servings):
        self.name = name
        self.servings = servings
        self.ingredients = {}
        pass

    def add_ingredient(self, ingredient, amount):
        if ingredient in self.ingredients:
            self.ingredients[ingredient] += amount
        else:
            self.ingredients[ingredient] = amount

    def scale(self, new_servings):
        factor = new_servings
        for ingredient in self.ingredients:
            self.ingredients[ingredient] *= factor

    def report(self):
        print(f"{self.name} for {self.servings} servings:")
        for ingredient, amount in self.ingredients.items():
            print(f"  {ingredient}: {amount}")
````

````python
## Do not modify this cell
dinner = Recipe("Spaghetti and meatballs", servings=4)
dinner.add_ingredient("spaghetti", 400)
dinner.add_ingredient("ground beef", 500)
dinner.add_ingredient("eggs", 1)
dinner.add_ingredient("ground beef", 100)
dinner.report()
````
