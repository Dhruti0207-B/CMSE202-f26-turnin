## 5: Creating a custom class to support your zoo (time permitting)



**&#9989; Do This:** Create your own class with at least three attributes and two methods to support the functions of your zoo.  You can use your imagination of a useful class to run a zoo. 
 
**Hints/suggestions:**  One suggestion is a class that can support the ticket sales at the zoo or another might be one that supports the feeding of the animals.  Or perhaps: consider a class that keeps track of the food costs for a specific animal or the overall cost of running the zoo.

Include methods to print out summary information.


```python
# put your code here
class FeedingPlan:
    def __init__(self, animal_kind, food_per_day_kg, cost_per_kg, days):
        self.animal_kind = animal_kind
        self.food_per_day_kg = food_per_day_kg
        self.cost_per_kg = cost_per_kg
        self.days = days
    def total_food(self):
        return self.food_per_day_kg * self.days
    def total_cost(self):
        return self.total_food() * self.cost_per_kg
    def summary(self):
        print("{0} feeding plan: {1:.1f} kg of food over {2} days, costing ${3:.2f}."
              .format(self.animal_kind, self.total_food(), self.days, self.total_cost()))
elephant_plan = FeedingPlan("elephant", 50, 2.50, 7)
elephant_plan.summary()
```
